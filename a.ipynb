## 1.2 Tell your computer what to do

As explained above, the programming involves an important element of coding, which is to tell the machine what to do.  On now day, the central processing unit (CPU) or graphic processing unit (GPU) are the *brain* of the computer. You will need to write computer code to be run by CPU/GPU, which contains basic instruction set – machine language – allowing users to communicate with CPU or GPU directly.  Such instruction sets are different for different CPU/GPU architectures, but they are the language that the computer can *understand*. For example, to ask the computer to add 3, if you are using Intel x86-64 architecture, you will need to use the Hexadecimal code 83 C0 03. The corresponding code for ARM architecture is 91000C00. It is not convenient to write such a code, although it is machine orientated. 

Alternatively, high-level languages have been developed to help make coding easier. Nevertheless, the computer hardware (CPU and GPU) cannot understand them directly. We need to place a **translator**, which converting the high-level languages into the machine code. There are two types of **translators**

- **Compilers** pre-translate codes into a set of low-level instructions, which is then executed. For example, c++ and Fortran
- **Interpreters** translate and execute code line by line. For example, Python and Matlab

Following the rapid development of Artificial Intelligence (AI), AI tools, e.g. Claude and Copilot, for assisting coding are now popular. It is understandable if you have the question like do we need to learn programming in the current world full of AI tools? If you look back to Section 1.1, you may understand that these tools may assist your coding, but you are responsible to choose correct formula and algorithms for the programming, which, in many cases, may be more important for engineering programming. Let’s look at the following example.

In an engineering practice,  the water depth at 4 different positions along a river have been measured. They are 0.4 m, 1.24 m, 1.30 m, 0.54 m for x = 0, 1 m, 2 m and 3 m, respectively.  I asked Copilot to write a python code to approximate the water depth at x = 1.5 m. The following is the code written by Copilot.


```python

import numpy as np

# Data
x = np.array([0, 1, 2, 3], dtype=float)
h = np.array([0.4, 1.24, 1.30, 0.54], dtype=float)

# Cubic polynomial through all four points
coeffs = np.polyfit(x, h, 3)
p = np.poly1d(coeffs)

# Evaluate at x = 1.5
x_query = 1.5
h_query = p(x_query)

print(f"Water depth at x = {x_query} is approximately {h_query:.6f}")
```

It works quite well. However, in engineering practices, we do not only need the approximation but also need to understand the accuracy and reliability of the approximation. To this end, we need to figure out the numerical algorithm behind the code and its degree of accuracy. For this purpose, you need to 
- understand the code to find the details of numerical algorithms that were implemented. You may notice from the comments, this code adopts cubic polynomial.  
- understand the algorithms to figure out how it works and what its degree of accuracy is. 
- assess whether this is an appropriate algorithm to meet the engineering requirement. 

All these require you to have a basic understanding of Python coding and commonly used numerical algorithms for engineering practices. We will go through these in the rest of the course.
